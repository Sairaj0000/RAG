```python
from langchain_community.document_loaders import PyPDFLoader
pdf_path="explanable recommendations using product knowledge graph.pdf"
loader=PyPDFLoader(pdf_path)
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50)
chunks=splitter.split_documents(docs)
chunks
```

```python
len(chunks)
```

```python
###Embeddings
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
embeddings.embed_query("what is AI")
```

```python
#create and stored the embeddings stored 
```

```python
from langchain_community.vectorstores import Chroma
vectorstore=Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="VectorDB"
)
```

```python
vectorstore.similarity_search("Endgame",k=4)
```

```python
##Talk ot LLM
```

```python
context=vectorstore.similarity_search("graph",k=4)
```

```python
response=llm.invoke(f"Graph:{context}")
print(response.content)
```
